```python
import matplotlib.pyplot as plt

top_cities = df['المدينة'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_cities.plot(kind='bar')
plt.title('Top 10 Cities by Number of Events')
plt.xlabel('City')
plt.ylabel('Number of Events')
plt.show()
```
